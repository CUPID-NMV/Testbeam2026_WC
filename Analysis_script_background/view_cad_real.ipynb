# CAD reale muon veto CUPID — contenitori + PMT (3D orientabile)

Dal file STEP dettagliato **`20260713-Assy-WC-Y.stp`** (3646 solidi, 48 Tank, 96 PMT).
- **41 contenitori** tassellati con la **geometria reale** -> si vedono i **fori veri** (cerchio sul fondo, slot sulle pareti).
- **96 PMT** (2 per tank, gruppo di solidi 20x68x68 identici) come marker neri, ai bordi di lettura dei tank.

Mesh in `cad/newcad_mesh.npz`, centri PMT in `cad/pmt_centers.npy`. Ruota/zooma col mouse.

L'acqua NON e' modellata: e' la cavita' interna del contenitore (bbox meno lo spessore parete ~10 mm),
che ricaviamo per il ray-tracing del percorso in acqua (passo successivo).

```python
import numpy as np
import plotly.graph_objects as go

CAD="/Users/benussi/Testbeam2026_WC_unified/cad"
z=np.load(f"{CAD}/newcad_mesh.npz"); V=z["V"]; F=z["F"]; C=z["C"]
P=np.load(f"{CAD}/pmt_centers.npy")
print(f"contenitori: vertici {len(V)} triangoli {len(F)}  |  PMT: {len(P)}")

COL={"Xp":"#d62728","Xm":"#ff9896","Yp":"#1f77b4","Ym":"#aec7e8","Zp":"#2ca02c","Zm":"#98df8a"}
LAB={"Xp":"X+","Xm":"X-","Yp":"Y+","Ym":"Y-","Zp":"Z+ (tetto)","Zm":"Z- (fondo)"}

fig=go.Figure()
for face,col in COL.items():
    m=C==face
    if not m.any(): continue
    Ff=F[m]
    fig.add_trace(go.Mesh3d(x=V[:,0],y=V[:,1],z=V[:,2],i=Ff[:,0],j=Ff[:,1],k=Ff[:,2],
        color=col,opacity=0.6,flatshading=True,hoverinfo="skip",name=LAB[face],showlegend=True))
fig.add_trace(go.Scatter3d(x=P[:,0],y=P[:,1],z=P[:,2],mode="markers",
    marker=dict(size=4,color="black"),name=f"PMT ({len(P)})"))
fig.update_layout(title="CUPID muon veto — CAD reale: contenitori (fori veri) + PMT",
    scene=dict(xaxis_title="X (mm)",yaxis_title="Y (mm)",zaxis_title="Z (mm)",aspectmode="data"),
    width=1000,height=850,legend=dict(title="faccia / PMT"))
fig.show()
```
